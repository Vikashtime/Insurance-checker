```python
from google.colab import files
from google.genai import types

# A box will appear. Click "Choose Files" and pick your photo.
uploaded = files.upload()
photo_name = list(uploaded.keys())[0]
photo_bytes = uploaded[photo_name]

# Tell Gemini what kind of file it is (change to image/png if your photo is a .png)
photo = types.Part.from_bytes(data=photo_bytes, mime_type="image/jpeg")

claim = "My kitchen floor got wet after a pipe burst on 12 Sep. Repair bill is Rs 84,000."

question = f"""You are an insurance helper. Read this claim:
{claim}

Also look at the attached photo.

Answer in exactly four lines:
WHAT I SEE: describe the photo in one sentence
DECISION: approve, reject, or needs human check
CONFIDENCE: a number from 0 to 100
REASON: one short sentence, mentioning both the message and the photo"""

answer = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=[photo, question]
)
print(answer.text)
```

```python
import json, time
from datetime import datetime
from google import genai
from google.genai import types
from google.colab import auth, files, userdata
from google.cloud import bigquery

# ---- CONNECT ----
auth.authenticate_user()
client = genai.Client(vertexai=True, api_key=userdata.get("MY_KEY"))
PROJECT = "ps42-bigquery"
bq = bigquery.Client(project=PROJECT)
bq.create_dataset("claims", exists_ok=True)

# ---- SETTINGS (change these if your theme changes) ----
ROLE = "You are an insurance claim helper."
MIN_CONFIDENCE = 75   # below this number, a human must check

claims = [
    {"id": "CLM-101", "filed": "01 Oct 2026",
     "text": "A pipe burst suddenly in my kitchen on 28 Sep 2026 and flooded the floor. The plumber bill is Rs 18,000."},
    {"id": "CLM-102", "filed": "01 Oct 2026",
     "text": "My phone worth Rs 40,000 was stolen from a cafe table on 29 Sep 2026 while I went to the counter."},
    {"id": "CLM-103", "filed": "01 Oct 2026",
     "text": "A leak under my bathroom sink damaged the cabinet on 1 Aug 2026. The repair cost is Rs 9,000."},
]

# ---- THE POLICY DOCUMENT ----
policy = types.Part.from_bytes(data=open("policy.pdf", "rb").read(), mime_type="application/pdf")

# ---- HELPER FUNCTIONS ----
def ask_photo(claim):
    print(f"\n>>> {claim['id']}: choose a photo, or press Cancel to skip.")
    up = files.upload()
    if not up:
        return None, None
    name = list(up.keys())[0]
    return name, up[name]

def find(lines, label):
    for line in lines:
        clean = line.replace("*", "").strip()
        if clean.upper().startswith(label):
            return clean.split(":", 1)[1].strip()
    return ""

def check_claim(claim, name, data):
    parts = [policy]
    if data:
        mime = "image/png" if name.lower().endswith(".png") else "image/jpeg"
        parts.append(types.Part.from_bytes(data=data, mime_type=mime))
        photo_note = "The second attachment is the claim photo."
    else:
        photo_note = "There is no photo for this claim."
    parts.append(f"""{ROLE}
The first attachment is the policy document. {photo_note}

Claim filed on: {claim['filed']}
Claim message: {claim['text']}

Check the claim against the policy clauses. Answer in exactly five lines:
WHAT I SEE: one sentence about the photo (or 'no photo')
DECISION: APPROVE, REJECT, or HUMAN CHECK
CONFIDENCE: a number from 0 to 100
CLAUSE: the clause number you used, like 9.4
REASON: one short sentence that mentions the clause""")
    resp = client.models.generate_content(model="gemini-2.5-flash", contents=parts)
    return resp.text

def safe_status(decision, confidence):
    d = decision.upper()
    if confidence < MIN_CONFIDENCE:
        return "HUMAN CHECK (low confidence)"
    if "APPROVE" in d:
        return "APPROVE"
    if "REJECT" in d:
        return "REJECT"
    return "HUMAN CHECK"

# ---- RUN EVERY CLAIM ----
results = []
for claim in claims:
    name, data = ask_photo(claim)
    try:
        text = check_claim(claim, name, data)
    except Exception as e:
        print("Problem with", claim["id"], ":", e)
        continue
    lines = text.split("\n")
    digits = "".join(c for c in find(lines, "CONFIDENCE") if c.isdigit())
    conf = int(digits) if digits else 0
    row = {
        "claim_id": claim["id"],
        "photo": name or "none",
        "what_i_see": find(lines, "WHAT I SEE"),
        "decision": find(lines, "DECISION"),
        "confidence": conf,
        "clause": find(lines, "CLAUSE"),
        "reason": find(lines, "REASON"),
        "final_status": safe_status(find(lines, "DECISION"), conf),
        "human_reviewed": "no",
        "created": datetime.now().isoformat(),
    }
    results.append(row)
    print(f"{row['claim_id']}: {row['final_status']} ({conf}%) - clause {row['clause']}")
    time.sleep(2)

# ---- SAVE TO BIGQUERY ----
if results:
    job = bq.load_table_from_json(results, f"{PROJECT}.claims.decisions_v2",
        job_config=bigquery.LoadJobConfig(autodetect=True, write_disposition="WRITE_APPEND"))
    job.result()
    print("\nSaved", len(results), "rows to BigQuery table claims.decisions_v2")

# ---- PRINT A COPY-PASTE BLOCK FOR THE SCREEN ----
print("\n===== COPY FROM HERE =====")
print(json.dumps(results, indent=1))
print("===== TO HERE =====")
```

```python
!pip install -q gradio
import gradio as gr
from datetime import datetime
from google import genai
from google.genai import types
from google.colab import auth, userdata
from google.cloud import bigquery

auth.authenticate_user()
client = genai.Client(vertexai=True, api_key=userdata.get("MY_KEY"))
PROJECT = "ps42-bigquery"
bq = bigquery.Client(project=PROJECT)
policy = types.Part.from_bytes(data=open("policy.pdf", "rb").read(), mime_type="application/pdf")
MIN_CONFIDENCE = 75

def find(lines, label):
    for line in lines:
        clean = line.replace("*", "").strip()
        if clean.upper().startswith(label):
            return clean.split(":", 1)[1].strip()
    return ""

def check(claim_text, filed, photo_path):
    if not claim_text.strip():
        return "Please type a claim first."
    parts = [policy]
    note = "There is no photo for this claim."
    photo_name = "none"
    if photo_path and photo_path.lower().endswith((".jpg", ".jpeg", ".png", ".webp")):
        ext = photo_path.lower().rsplit(".", 1)[1]
        mime = {"jpg": "image/jpeg", "jpeg": "image/jpeg", "png": "image/png", "webp": "image/webp"}[ext]
        parts.append(types.Part.from_bytes(data=open(photo_path, "rb").read(), mime_type=mime))
        note = "The second attachment is the claim photo."
        photo_name = photo_path.split("/")[-1]
    parts.append(f"""You are an insurance claim helper.
The first attachment is the policy document. {note}
Claim filed on: {filed}
Claim message: {claim_text}

Check the claim against the policy clauses. Answer in exactly five lines:
WHAT I SEE: one sentence about the photo (or 'no photo')
DECISION: APPROVE, REJECT, or HUMAN CHECK
CONFIDENCE: a number from 0 to 100
CLAUSE: the clause number you used, like 9.4
REASON: one short sentence that mentions the clause""")
    try:
        text = client.models.generate_content(model="gemini-2.5-flash", contents=parts).text
    except Exception as e:
        return f"Something went wrong: {e}"
    lines = text.split("\n")
    digits = "".join(c for c in find(lines, "CONFIDENCE") if c.isdigit())
    conf = int(digits) if digits else 0
    decision = find(lines, "DECISION")
    final = "HUMAN CHECK (low confidence)" if conf < MIN_CONFIDENCE else ("APPROVE" if "APPROVE" in decision.upper() else "REJECT" if "REJECT" in decision.upper() else "HUMAN CHECK")
    row = {"claim_id": "LIVE-" + datetime.now().strftime("%H%M%S"), "photo": photo_name,
           "what_i_see": find(lines, "WHAT I SEE"), "decision": decision, "confidence": conf,
           "clause": find(lines, "CLAUSE"), "reason": find(lines, "REASON"),
           "final_status": final, "human_reviewed": "no", "created": datetime.now().isoformat()}
    try:
        bq.load_table_from_json([row], f"{PROJECT}.claims.decisions_v2",
            job_config=bigquery.LoadJobConfig(autodetect=True, write_disposition="WRITE_APPEND")).result()
        saved = "Saved to BigQuery."
    except Exception as e:
        saved = f"Could not save to BigQuery: {e}"
    return f"""## {final}
**Confidence:** {conf}%  |  **Clause used:** {row['clause']}

**What Gemini saw:** {row['what_i_see']}

**Reason:** {row['reason']}

*{saved}*"""

demo = gr.Interface(
    fn=check,
    inputs=[gr.Textbox(label="Claim message", lines=4),
            gr.Textbox(label="Date filed", value=datetime.now().strftime("%d %b %Y")),
            gr.Image(type="filepath", label="Photo (optional)")],
    outputs=gr.Markdown(),
    title="Insurance Claim Checker (PS42)",
    description="Gemini checks your claim against the policy and explains its decision.")
demo.launch(share=True)
```
